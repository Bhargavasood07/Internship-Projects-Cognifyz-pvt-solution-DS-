```python
import pandas as pd
```

```python
df = pd.read_csv("/content/Dataset .csv")
```

```python
print("Original Dataset:")
print(df.head())
```

```python
df['Restaurant Name Length'] = (
    df['Restaurant Name'].str.len()
)

print("\nRestaurant Name Length:")
print(df[['Restaurant Name',
          'Restaurant Name Length']].head())
```

```python

df['Address Length'] = df['Address'].str.len()

print("\nAddress Length:")
print(df[['Address', 'Address Length']].head())
```

```python
df['Has Table Booking'] = (
    df['Has Table booking'].map({
        'Yes': 1,
        'No': 0
    })
)

print("\nTable Booking Feature:")
print(df['Has Table Booking'].value_counts())
```

```python
df['Has Online Delivery'] = (
    df['Has Online delivery'].map({
        'Yes': 1,
        'No': 0
    })
)

print("\nOnline Delivery Feature:")
print(df['Has Online Delivery'].value_counts())
```

```python

print("\nNewly Created Features:")

print(df[[
    'Restaurant Name',
    'Restaurant Name Length',
    'Address Length',
    'Has Table Booking',
    'Has Online Delivery'
]].head(10))
```

```python
print("\nMissing Values:")
print(df[[
    'Restaurant Name Length',
    'Address Length',
    'Has Table Booking',
    'Has Online Delivery'
]].isnull().sum())
```

```python
df.to_csv("Feature_Engineered_Dataset.csv",
          index=False)

print("\nFeature Engineering Completed Successfully!")

print("Updated Dataset Shape:", df.shape)
print("File Saved: Feature_Engineered_Dataset.csv")
```
